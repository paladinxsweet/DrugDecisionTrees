```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import preprocessing
from sklearn.model_selection import train_test_split 
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
from sklearn.tree import export_graphviz
```

```python
my_data = pd.read_csv('https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-ML0101EN-SkillsNetwork/labs/Module%203/data/drug200.csv', delimiter=",")
my_data.head()
```

```python
my_data.shape
```

```python
my_data.isnull().sum()
```

```python
my_data.columns
```

```python
X=my_data[['Age', 'Sex', 'BP', 'Cholesterol', 'Na_to_K']].values
X[0:5]
```

```python
print(my_data['Sex'].unique())
print(my_data['BP'].unique())
print(my_data['Cholesterol'].unique())
```

## PRE_PROCESSING

```python
le_sex = preprocessing.LabelEncoder()
le_sex.fit(['F','M'])
X[:,1] = le_sex.transform(X[:,1]) 


le_BP = preprocessing.LabelEncoder()
le_BP.fit([ 'LOW', 'NORMAL', 'HIGH'])
X[:,2] = le_BP.transform(X[:,2])


le_Chol = preprocessing.LabelEncoder()
le_Chol.fit([ 'NORMAL', 'HIGH'])
X[:,3] = le_Chol.transform(X[:,3]) 

X[0:5]
```

```python
y=my_data['Drug']
```

```python
y[0:5]
```

## Setting up the Decision Tree

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=3)
```

```python
print(f"X train shape:{X_train.shape}\nY train shape:{y_train.shape}")
print(f"X test shape{X_test.shape}\nY test shape:{y_test.shape}")
```

## Modeling

```python
drugTree=DecisionTreeClassifier(criterion="entropy",max_depth=4)
drugTree
```

```python
drugTree.fit(X_train,y_train)
```

## Prediction

```python
predTree=drugTree.predict(X_test)
```

```python
print(predTree[0:5])
print(y_test[0:5])
```

## Evaluation

```python
print("DecisionTrees's Accuracy:",accuracy_score(y_test,predTree))
```
